```python

from pathlib import Path
import subprocess
import shutil

RAR_FILE = Path("/content/archive.rar")
EXTRACT_DIR = Path("/content/underwater_dataset_extracted")
EXTRACT_DIR.mkdir(parents=True, exist_ok=True)

if not RAR_FILE.exists():
    raise FileNotFoundError(
        "archive.rar not found. Upload your RAR file again."
    )

# Try the available extractors
extractors = []

if shutil.which("unrar"):
    extractors.append(["unrar", "x", "-o+", str(RAR_FILE), str(EXTRACT_DIR) + "/"])

if shutil.which("7z"):
    extractors.append(["7z", "x", "-y", f"-o{EXTRACT_DIR}", str(RAR_FILE)])

if shutil.which("7zz"):
    extractors.append(["7zz", "x", "-y", f"-o{EXTRACT_DIR}", str(RAR_FILE)])

success = False

for cmd in extractors:
    print("Trying:", " ".join(cmd[:2]))

    result = subprocess.run(
        cmd,
        capture_output=True,
        text=True
    )

    if result.returncode == 0:
        success = True
        print(result.stdout[-2000:])
        break

    print("Extraction failed:")
    print(result.stderr[-2000:] or result.stdout[-2000:])

if not success:
    raise RuntimeError(
        "Extraction failed with all available tools. "
        "Check the diagnostic output above."
    )

print("\nExtraction completed!")
print("Files and folders:", [
    p.name for p in EXTRACT_DIR.iterdir()
][:30])
```

```python
import os, random, copy, time
from pathlib import Path
import numpy as np, pandas as pd, matplotlib.pyplot as plt
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms, models
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix, classification_report
from tqdm.auto import tqdm

SEED=42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
if torch.cuda.is_available(): torch.cuda.manual_seed_all(SEED)
DEVICE=torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE, torch.cuda.get_device_name(0) if DEVICE.type=="cuda" else "")
```

```python
# Configuration for Google Colab
# EXTRACT_DIR is created by the upload/extraction cell above.
IMAGE_SIZE = 224
BATCH_SIZE = 8
EPOCHS = 35
LR = 1e-4
WEIGHT_DECAY = 1e-4
VAL_SIZE = 0.20
TEST_SIZE = 0.15
NUM_WORKERS = 2 if DEVICE.type == "cuda" else 0
USE_PRETRAINED = True
OUT = Path("/content/hybrid_attention_results")
OUT.mkdir(exist_ok=True, parents=True)

# Find the directory containing matching train/ and test/ folders.
def find_dataset_root(root):
    root = Path(root)
    for train_dir in [p for p in root.rglob("*") if p.is_dir() and p.name.lower() in {"train", "training"}]:
        parent = train_dir.parent
        test_dirs = [p for p in parent.iterdir() if p.is_dir() and p.name.lower() in {"test", "testing"}]
        if test_dirs:
            return parent
    # If archive contains class folders directly, return the directory with class subfolders.
    for folder in [root] + [p for p in root.rglob("*") if p.is_dir()]:
        subdirs = [p for p in folder.iterdir() if p.is_dir()]
        if len(subdirs) >= 2 and any(
            (p / "train").is_dir() or (p / "training").is_dir() for p in subdirs
        ):
            continue
        if len(subdirs) >= 2 and not any(p.name.lower() in {"train", "test", "training", "testing"} for p in subdirs):
            image_exts = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
            if any(f.is_file() and f.suffix.lower() in image_exts for d in subdirs for f in list(d.iterdir())[:5]):
                return folder
    raise FileNotFoundError(
        "Could not locate the dataset. Expected train/class_name and test/class_name folders "
        "inside the extracted archive. Inspect EXTRACT_DIR and adjust the root if needed."
    )

DATASET_DIR = find_dataset_root(EXTRACT_DIR)
print("Dataset:", DATASET_DIR.resolve(), "| exists:", DATASET_DIR.exists())
print("Output:", OUT.resolve())
```

```python
# Discover train/test folders and build stratified splits
def find_named_dir(root, accepted_names):
    for p in [Path(root)] + [q for q in Path(root).rglob("*") if q.is_dir()]:
        if p.is_dir() and p.name.lower() in accepted_names:
            # A valid split directory should contain class subfolders.
            if any(q.is_dir() for q in p.iterdir()):
                return p
    return None

train_dir = find_named_dir(DATASET_DIR, {"train", "training"})
test_dir = find_named_dir(DATASET_DIR, {"test", "testing"})

if train_dir is not None:
    base_train = datasets.ImageFolder(train_dir)
    class_names = base_train.classes
    if len(class_names) < 2:
        raise ValueError(f"Expected at least two class folders in {train_dir}")
    targets = np.asarray(base_train.targets)
    indices = np.arange(len(targets))
    if len(indices) < 5:
        raise ValueError("Too few training images to create train/validation splits.")
    train_idx, val_idx = train_test_split(
        indices, test_size=VAL_SIZE, random_state=SEED, stratify=targets
    )
    # Separate ImageFolder instances are important because train and validation use different transforms.
    train_dataset = Subset(datasets.ImageFolder(train_dir), train_idx.tolist())
    val_dataset = Subset(datasets.ImageFolder(train_dir), val_idx.tolist())

    if test_dir is not None:
        test_dataset = datasets.ImageFolder(test_dir)
        if test_dataset.classes != class_names:
            raise ValueError(
                f"Train/test class folders differ. Train={class_names}; test={test_dataset.classes}"
            )
    else:
        print("WARNING: No separate test folder found; making train/validation/test splits from one folder.")
        trval_idx, test_idx = train_test_split(
            indices, test_size=TEST_SIZE, random_state=SEED, stratify=targets
        )
        train_idx, val_idx = train_test_split(
            trval_idx, test_size=VAL_SIZE, random_state=SEED, stratify=targets[trval_idx]
        )
        train_dataset = Subset(datasets.ImageFolder(train_dir), train_idx.tolist())
        val_dataset = Subset(datasets.ImageFolder(train_dir), val_idx.tolist())
        test_dataset = Subset(datasets.ImageFolder(train_dir), test_idx.tolist())
else:
    # If there is no train folder, expect class folders directly under DATASET_DIR.
    full = datasets.ImageFolder(DATASET_DIR)
    class_names = full.classes
    if len(class_names) < 2:
        raise ValueError("No train folder found and dataset root must contain class subfolders.")
    targets = np.asarray(full.targets)
    indices = np.arange(len(targets))
    trval_idx, test_idx = train_test_split(
        indices, test_size=TEST_SIZE, random_state=SEED, stratify=targets
    )
    train_idx, val_idx = train_test_split(
        trval_idx, test_size=VAL_SIZE, random_state=SEED, stratify=targets[trval_idx]
    )
    train_dataset = Subset(datasets.ImageFolder(DATASET_DIR), train_idx.tolist())
    val_dataset = Subset(datasets.ImageFolder(DATASET_DIR), val_idx.tolist())
    test_dataset = Subset(datasets.ImageFolder(DATASET_DIR), test_idx.tolist())

NUM_CLASSES = len(class_names)
print("Classes:", class_names)
print("Train/validation/test:", len(train_dataset), len(val_dataset), len(test_dataset))
if len(train_dataset) == 0 or len(val_dataset) == 0 or len(test_dataset) == 0:
    raise ValueError("A dataset split is empty. Check the archive and split sizes.")
```

```python
# Transforms and data loaders
mean=(0.485,0.456,0.406); std=(0.229,0.224,0.225)
train_tf=transforms.Compose([
    transforms.Resize((IMAGE_SIZE,IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=.12, contrast=.12, saturation=.10, hue=.02),
    transforms.ToTensor(), transforms.Normalize(mean,std)])
eval_tf=transforms.Compose([
    transforms.Resize((IMAGE_SIZE,IMAGE_SIZE)),
    transforms.ToTensor(), transforms.Normalize(mean,std)])

def set_tf(ds, tf):
    (ds.dataset if isinstance(ds,Subset) else ds).transform=tf
set_tf(train_dataset,train_tf); set_tf(val_dataset,eval_tf); set_tf(test_dataset,eval_tf)
train_loader=DataLoader(train_dataset,batch_size=BATCH_SIZE,shuffle=True,num_workers=NUM_WORKERS,pin_memory=DEVICE.type=="cuda")
val_loader=DataLoader(val_dataset,batch_size=BATCH_SIZE,shuffle=False,num_workers=NUM_WORKERS,pin_memory=DEVICE.type=="cuda")
test_loader=DataLoader(test_dataset,batch_size=BATCH_SIZE,shuffle=False,num_workers=NUM_WORKERS,pin_memory=DEVICE.type=="cuda")
print("Loaders ready.")
```

```python
class ChannelAttention(nn.Module):
    def __init__(self,c,reduction=16):
        super().__init__(); h=max(c//reduction,8)
        self.mlp=nn.Sequential(nn.Conv2d(c,h,1,bias=False),nn.ReLU(inplace=True),nn.Conv2d(h,c,1,bias=False))
        self.avg=nn.AdaptiveAvgPool2d(1); self.max=nn.AdaptiveMaxPool2d(1)
    def forward(self,x):
        return x*torch.sigmoid(self.mlp(self.avg(x))+self.mlp(self.max(x)))

class SpatialAttention(nn.Module):
    def __init__(self):
        super().__init__(); self.conv=nn.Conv2d(2,1,7,padding=3,bias=False)
    def forward(self,x):
        a=x.mean(1,keepdim=True); m=x.max(1,keepdim=True).values
        return x*torch.sigmoid(self.conv(torch.cat([a,m],1)))

class HybridAttentionNet(nn.Module):
    def __init__(self,num_classes,pretrained=True,fusion_channels=128,dropout=.4):
        super().__init__()
        ew=models.EfficientNet_V2_S_Weights.DEFAULT if pretrained else None
        cw=models.ConvNeXt_Tiny_Weights.DEFAULT if pretrained else None
        eff=models.efficientnet_v2_s(weights=ew); cnx=models.convnext_tiny(weights=cw)
        self.eff_features=eff.features; self.cnx_features=cnx.features
        self.eff_proj=nn.Sequential(nn.Conv2d(1280,fusion_channels,1,bias=False),nn.BatchNorm2d(fusion_channels),nn.GELU())
        self.cnx_proj=nn.Sequential(nn.Conv2d(768,fusion_channels,1,bias=False),nn.BatchNorm2d(fusion_channels),nn.GELU())
        self.fusion=nn.Sequential(
            nn.Conv2d(2*fusion_channels,2*fusion_channels,3,padding=1,bias=False),
            nn.BatchNorm2d(2*fusion_channels),nn.GELU(),
            nn.Conv2d(2*fusion_channels,fusion_channels,1,bias=False),
            nn.BatchNorm2d(fusion_channels),nn.GELU())
        self.ca=ChannelAttention(fusion_channels); self.sa=SpatialAttention()
        self.refine=nn.Sequential(nn.Conv2d(fusion_channels,fusion_channels,3,padding=1,bias=False),
                                  nn.BatchNorm2d(fusion_channels),nn.GELU())
        self.pool=nn.AdaptiveAvgPool2d(1)
        self.head=nn.Sequential(nn.Flatten(),nn.Linear(fusion_channels,256),nn.GELU(),
                                nn.Dropout(dropout),nn.Linear(256,num_classes))
    def forward(self,x):
        a=self.eff_proj(self.eff_features(x)); b=self.cnx_proj(self.cnx_features(x))
        if a.shape[-2:]!=b.shape[-2:]:
            b=F.interpolate(b,size=a.shape[-2:],mode="bilinear",align_corners=False)
        z=self.fusion(torch.cat([a,b],dim=1))
        z=self.ca(z); z=self.sa(z); z=z+self.refine(z)
        return self.head(self.pool(z))

model=HybridAttentionNet(NUM_CLASSES,pretrained=USE_PRETRAINED).to(DEVICE)
print("Trainable parameters:",f"{sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
with torch.no_grad():
    xb,_=next(iter(train_loader))
    print("Output shape:",tuple(model(xb[:1].to(DEVICE)).shape))
```

```python
# Training utilities
criterion=nn.CrossEntropyLoss()
optimizer=torch.optim.AdamW(model.parameters(),lr=LR,weight_decay=WEIGHT_DECAY)
scheduler=torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer,mode="min",factor=.5,patience=2)
amp_enabled=DEVICE.type=="cuda"
scaler=torch.amp.GradScaler("cuda",enabled=amp_enabled)

def run_epoch(loader,training=False):
    model.train(training); total=0; loss_sum=0.; ys=[]; ps=[]
    with (torch.enable_grad() if training else torch.no_grad()):
        for x,y in tqdm(loader,leave=False):
            x=x.to(DEVICE); y=y.to(DEVICE)
            if training: optimizer.zero_grad(set_to_none=True)
            with torch.autocast(device_type=DEVICE.type,enabled=amp_enabled):
                logits=model(x); loss=criterion(logits,y)
            if training:
                scaler.scale(loss).backward(); scaler.step(optimizer); scaler.update()
            loss_sum+=loss.item()*y.size(0); total+=y.size(0)
            ys.extend(y.detach().cpu().tolist()); ps.extend(logits.argmax(1).detach().cpu().tolist())
    return loss_sum/max(total,1), accuracy_score(ys,ps), ys, ps

print("Training functions ready.")
```

```python
# Train and save best checkpoint by validation loss
best_loss=float("inf"); best_state=None; patience=5; bad_epochs=0; history=[]
for epoch in range(1,EPOCHS+1):
    t=time.time()
    tr_loss,tr_acc,_,_=run_epoch(train_loader,True)
    va_loss,va_acc,_,_=run_epoch(val_loader,False)
    scheduler.step(va_loss)
    row={"epoch":epoch,"train_loss":tr_loss,"train_accuracy":tr_acc,
         "val_loss":va_loss,"val_accuracy":va_acc,"lr":optimizer.param_groups[0]["lr"],
         "seconds":time.time()-t}
    history.append(row); pd.DataFrame(history).to_csv(OUT/"training_history.csv",index=False)
    print(f"Epoch {epoch}/{EPOCHS} | train loss={tr_loss:.4f} acc={tr_acc:.4f} | val loss={va_loss:.4f} acc={va_acc:.4f} | {row['seconds']:.1f}s")
    if va_loss<best_loss:
        best_loss=va_loss; best_state=copy.deepcopy(model.state_dict()); bad_epochs=0
        torch.save({"model_state_dict":best_state,"class_names":class_names,"num_classes":NUM_CLASSES,
                    "image_size":IMAGE_SIZE,"seed":SEED,
                    "architecture":"EfficientNetV2-S + ConvNeXt-Tiny + feature fusion + channel/spatial attention"},
                   OUT/"best_hybrid_attention_model.pth")
        print("  Best checkpoint saved.")
    else:
        bad_epochs+=1
        if bad_epochs>=patience:
            print("Early stopping."); break
if best_state is not None: model.load_state_dict(best_state)
print("Training finished. Best validation loss:",best_loss)
```

```python
# Training curves
hist=pd.DataFrame(history); display(hist)
fig,ax=plt.subplots(1,2,figsize=(12,4))
ax[0].plot(hist.epoch,hist.train_loss,label="Train"); ax[0].plot(hist.epoch,hist.val_loss,label="Validation")
ax[0].set_title("Loss"); ax[0].set_xlabel("Epoch"); ax[0].legend()
ax[1].plot(hist.epoch,hist.train_accuracy,label="Train"); ax[1].plot(hist.epoch,hist.val_accuracy,label="Validation")
ax[1].set_title("Accuracy"); ax[1].set_xlabel("Epoch"); ax[1].legend()
plt.tight_layout(); plt.savefig(OUT/"training_curves.png",dpi=200); plt.show()
```

```python
# Final test-set evaluation, metrics, inference time, and confusion matrix
model.eval()
test_loss, test_acc, y_true, y_pred = run_epoch(test_loader, False)

# Measure average forward-pass time (excluding the first warm-up pass).
latencies = []
with torch.no_grad():
    for x, _ in test_loader:
        x = x.to(DEVICE)
        if DEVICE.type == "cuda":
            torch.cuda.synchronize()
        start = time.perf_counter()
        with torch.autocast(device_type=DEVICE.type, enabled=amp_enabled):
            _ = model(x)
        if DEVICE.type == "cuda":
            torch.cuda.synchronize()
        latencies.append((time.perf_counter() - start) / max(1, x.size(0)))
inference_ms_per_image = float(np.mean(latencies) * 1000) if latencies else float("nan")

macro_p, macro_r, macro_f1, _ = precision_recall_fscore_support(
    y_true, y_pred, average="macro", zero_division=0
)
weighted_p, weighted_r, weighted_f1, _ = precision_recall_fscore_support(
    y_true, y_pred, average="weighted", zero_division=0
)
print(
    f"Test loss: {test_loss:.4f}\nAccuracy: {test_acc:.4f}\n"
    f"Macro precision: {macro_p:.4f}\nMacro recall: {macro_r:.4f}\nMacro F1: {macro_f1:.4f}\n"
    f"Weighted precision: {weighted_p:.4f}\nWeighted recall: {weighted_r:.4f}\n"
    f"Weighted F1: {weighted_f1:.4f}\nApprox. inference time: {inference_ms_per_image:.2f} ms/image"
)

report = classification_report(
    y_true, y_pred, target_names=class_names, zero_division=0, output_dict=True
)
pd.DataFrame(report).transpose().to_csv(OUT / "test_classification_report.csv")

summary = pd.DataFrame([{
    "model": "HybridAttentionNet",
    "test_loss": test_loss,
    "test_accuracy": test_acc,
    "macro_precision": macro_p,
    "macro_recall": macro_r,
    "macro_f1": macro_f1,
    "weighted_precision": weighted_p,
    "weighted_recall": weighted_r,
    "weighted_f1": weighted_f1,
    "inference_ms_per_image": inference_ms_per_image,
    "num_classes": NUM_CLASSES,
    "test_samples": len(test_dataset)
}])
display(summary)
summary.to_csv(OUT / "test_summary.csv", index=False)

cm = confusion_matrix(y_true, y_pred, labels=list(range(NUM_CLASSES)))
fig, ax = plt.subplots(figsize=(max(8, NUM_CLASSES * .8), max(6, NUM_CLASSES * .6)))
im = ax.imshow(cm, interpolation="nearest")
ax.set_title("Test Confusion Matrix")
ax.set_xlabel("Predicted")
ax.set_ylabel("True")
ax.set_xticks(range(NUM_CLASSES))
ax.set_yticks(range(NUM_CLASSES))
ax.set_xticklabels(class_names, rotation=45, ha="right")
ax.set_yticklabels(class_names)
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(j, i, str(cm[i, j]), ha="center", va="center")
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.savefig(OUT / "confusion_matrix.png", dpi=200)
plt.show()

pd.DataFrame({
    "true_index": y_true,
    "predicted_index": y_pred,
    "true_class": [class_names[i] for i in y_true],
    "predicted_class": [class_names[i] for i in y_pred],
    "correct": [a == b for a, b in zip(y_true, y_pred)]
}).to_csv(OUT / "test_predictions.csv", index=False)

print("All outputs saved in:", OUT.resolve())
```
